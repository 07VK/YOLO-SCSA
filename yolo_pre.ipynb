```python
from ultralytics import YOLO
YAML_PATH = "/home/vfk5129/datasets/data_local.yaml"
```

```python
model_path = "/home/vfk5129/project/yolo_scsa/ultralytics/cfg/models/v8/yolov8s_SCSA_CWM.yaml"
model = YOLO(model_path)
model.train(data=YAML_PATH, imgsz=512, epochs=100, name='SCSA(s)', project='./logs', save=True)
```

```python
check_path = "/home/vfk5129/project/yolo_scsa/logs/SCSA(s)/weights/best.pt"
model = YOLO(check_path)
```

```python
model.val(data=YAML_PATH, imgsz=512)
```

```python
results = model.predict(source="/home/vfk5129/datasets/test/images", imgsz=512)
```

```python
for i, result in enumerate(results):
    result.save()
```

```python
import pandas as pd

for i, result in enumerate(results):
    # Access bounding boxes from the result
    boxes = result.boxes  # Access boxes attribute

    if boxes is not None and len(boxes.xyxy) > 0:  # Check if there are any boxes detected
        # Transfer the tensor to CPU and convert to NumPy array
        boxes_array = boxes.xyxy.cpu().numpy()  # Bounding box coordinates
        confidences = boxes.conf.cpu().numpy()  # Confidence scores
        classes = boxes.cls.cpu().numpy()  # Class labels

        # Combine all into a single DataFrame
        df = pd.DataFrame(
            data={
                "x1": boxes_array[:, 0],
                "y1": boxes_array[:, 1],
                "x2": boxes_array[:, 2],
                "y2": boxes_array[:, 3],
                "confidence": confidences,
                "class": classes
            }
        )

        print(f"Detections for image {i}:\n", df)  # Print detected results
    else:
        print(f"No detections found in image {i}.")
```

```python
metrics = results.results_dict
for metric_name, metric_value in metrics.items():
    print(f"{metric_name}: {metric_value}")
```
